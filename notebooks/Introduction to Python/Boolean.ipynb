# Boolean
A **boolean** is an object that can assume only two values: *True* or *False*

````python
t = True
f = False
print("t:", t, "type:", type(t))
print("f:", f, "type:", type(f))
````

````python
# From numbers to booleans
t = bool(1)                           # integer 1 or float 1.0 --> True
print("t:", t, "type:", type(t))

f = bool(0)                           # integer 0 or float 0.0 --> False
print("f:", f, "type:", type(f))
````

**Comparison**, **Logical**, and **bitwise logical** operations evaluate **boolean** *True* or *False*.
- **Comparison operation**  

```>``` : greater than

```<``` : less than

```>=``` : greater than or equal to

```<=``` : less than or equal to

```==``` : equal

```!=``` : not equal  

- **Logical operators**

Logical operators evaluate *True* or *False*

```and``` : and

```or``` : or

```not``` : not

- **Bitwise logical operators**  

```&``` : and

```|``` : or

```~``` : not

```^``` : xor

````python
# Comparison operations examples
print("5 > 1?", 5 > 1)
print("5 < 1?", 5 < 1)
print("5 >= 1?", 5 >= 1)
print("5 <= 1?", 5 <= 1)
print("5 == 1?", 5 == 1) # one = is an assignment; two == is for comparison
print("5 != 1?", 5 != 1)
````

````python
# Logical operations examples
print("True AND True:", True and True)
print("True OR False:", True or False)
print("NOT False:", not False)
````

````python
# Bitwise logical operations examples
print("0 & 0:", 0 & 0)
print("1 | 0:", 1 | 0)
print("1 ^ 0:", 1 ^ 0)
````
