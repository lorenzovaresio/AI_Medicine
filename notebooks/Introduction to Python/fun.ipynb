# Functions

A **function** is a reusable block of code that runs **only when it is called**.  
Functions help organize code, avoid repetition, and improve readability.

To use a function, you first **define** it with the `def` keyword, and then **call** it when needed.
```python
# Function definition
def func_name(par1, par2, par3, ...):
    instruction
    return out1, out2, oput3, ...

# Function call
out1,out2, ou3, ... = func_name(par1, par2, par3, ...)
```

- Parameters (par1, par2...) are the **inputs**
- The `return` statement sends back the **outputs**

````python
def square(x):
    y = x**2
    return y

var = 2
out = square(var)
print(out)
````

## A Warning About Parameters

In Python, function parameters are **passed by reference**:  
this means the function **receives a reference to the actual object**, not a copy.

- If the parameter is an **immutable** object (like a number, string, or tuple), the function cannot change the original value.
- If the parameter is a **mutable** object (like a list or dictionary), the function **can modify** it — and the changes will affect the original object **outside** the function.

This behavior is powerful, but requires attention:  
**changes inside a function can persist after the function call**, depending on the data type.

### Example: mutable vs. immutable parameters

Let's see the difference between how functions handle **immutable** and **mutable** objects.



````python
### Immutable (e.g. integer)

def modify_number(x):
  x = x + 1
  print("Inside function:", x)

a = 5
modify_number(a)
print("Outside function:", a)  # a is still 5
````

Even though `x` was modified inside the function, the original variable `a` is **unchanged** — integers are immutable.



````python
### Mutable (e.g. list)

def modify_list(lst):
  lst.append(4)
  print("Inside function:", lst)

my_list = [1, 2, 3]
modify_list(my_list)
print("Outside function:", my_list)  # my_list is changed
````

In this case, the function modified the original list — because lists are **mutable**.

---

🧠 **Tip**: Be careful when passing mutable objects to functions.  

## Function parameters and default values

In the **definition phase**, you can add as many parameters as you want — just separate them with commas.  
You can also define **default values** for some parameters by assigning a value in the function header.

In the **call phase**:
- Parameters **with default values** are **optional**
- Parameters **without default values** are **mandatory** and must be passed either **by position** or **by name**

````python
# Function definition with default parameter
def pow(base, exp = 1):
    out = base**exp
    return out
````

This function raises `base` to the power of `exp`.  
If `exp` is not specified, it defaults to 1.

````python
# Function call – mandatory parameter by position, default used
var = 2
y = pow(var,2)
print(y)
````

Only the base is provided (`var = 2`), so the function uses the default exponent `1`.

````python
# Function Call – both parameters by position
y = pow(var, 3)
print(y)
````

Both `base` and `exp` are passed by position: `2^3 = 8`.

````python
# Function call – mandatory parameter by name
y = pow(base = var)
print(y)
````

The base is passed by name; the exponent is omitted and defaults to `1`.

````python
# Function Call – both parameters by name
y = pow(base = var, exp = 3)
print(y)
````

Here, both arguments are passed **by name**.

````python
# Function call – order of named parameters can be switched
y = pow(exp = 3, base = var)
print(y)
````

When using **named arguments**, the order doesn't matter.

````python
# Function call – invalid keyword argument (TypeError)
y = pow(exp = 3, x = var)
print(y)
````

This will raise a `TypeError` because `x` is **not a valid parameter** of the `pow()` function.

````python
### Exercise 1
# Write a Python function to find the maximum of three numbers
def max3 (num1, num2, num3):
  max = num1
  if num2 > max
    max = num2
  if num3 > max
    max = num3
  return max
````
