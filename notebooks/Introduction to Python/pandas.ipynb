# pandas

[pandas](https://pandas.pydata.org/docs/index.html#module-pandas) is an open-source, BSD-licensed library that provides **high-performance**, **easy-to-use** data structures and tools for **data analysis** in Python.

This is a short introduction to **pandas**, intended for new users.  
For more advanced examples, refer to the official [Cookbook](https://pandas.pydata.org/pandas-docs/stable/user_guide/cookbook.html#cookbook).

To use pandas, we first need to **import** the module:

````python
# Importing pandas
import pandas as pd
````

The module is typically imported as `pd` by convention.

## pandas DataFrame

pandas provides two primary classes for working with data:

1. **Series** – a one-dimensional labeled array (like a column)
2. **DataFrame** – a two-dimensional table with rows and columns (like an Excel spreadsheet)

We can create a `DataFrame` by passing a **list of dictionaries**,  
where each dictionary represents a row, and the keys are column names.

````python
# Example: List of Dictionaries as Data
data = [{'ID': '001', 'Age': 68, 'Sex':'M', 'CT':2.543}, {'Age': 56, 'ID': '002', 'Sex': 'F', 'CT':2.34}, {'Age': 56, 'ID': '003', 'Sex': 'F', 'CT':2.41}]
print (type(data))
print (type(data[0]))
````

````python
# Creating a DataFrame
df = pd.DataFrame(data)
print (df)
````

This creates a `DataFrame` where each row comes from one dictionary in the list.  
The column names are automatically inferred from the keys.

### Column data types

The columns of a DataFrame can have **different data types**.

By default:
- Numbers are stored as `int` or `float`
- Strings are stored as `object` dtype (which can hold any type, including strings)

You can convert object columns to actual string type (`str`) if needed (df['ColumnName'] = df['ColumnName'].astype('string') )

````python
# Checking Data Types of Each Column
df.dtypes
````

This will display the data type (`dtype`) of each column in the DataFrame.

````python
# View Column Names
df.columns
````

The `.shape` attribute returns a **tuple** representing the dimensions of the DataFrame:

````python
# View DataFrame Shape
df.shape
````

### Selecting a Column

You can access a single column of a DataFrame using **dot notation** (`df.ColumnName`) or **bracket notation** (`df['ColumnName']`).

This returns a `Series` object — a one-dimensional array-like structure.

````python
# Example: Access the 'CT' Column
CT_dataset = df.CT
print (CT_dataset)
print (type(CT_dataset))
````

The result is a pandas `Series` containing all the values from the `'CT'` column.

## Selection

You can select specific data from a DataFrame using **labels** or **conditions**.

For detailed reference, see:
- [Indexing and Selecting Data](https://pandas.pydata.org/pandas-docs/stable/user_guide/indexing.html#indexing)

Passing a **column label** (string) in brackets selects that column and returns a `Series`.

This is equivalent to using dot notation (`df.Age`), but bracket notation is preferred when column names have spaces or special characters.

### Selecting rows by slice

For a DataFrame, using slicing syntax like `start:stop` selects **rows by position**,  
where `stop` is **excluded** (just like with Python lists).

````python
# Example: Select First Two Rows
df[0:2]
````

# Selection by Position with iloc

You can use `.iloc[]` to select rows and columns **by integer position**.  
For more details, see the [Selection by Position documentation](https://pandas.pydata.org/pandas-docs/stable/user_guide/indexing.html#indexing-integer).

````python
# Example: Select Row at Position 2
df.iloc[2]
````

This returns the **third row** (index starts at 0) as a Series.

# Integer slicing with iloc

Integer slicing with `.iloc[]` behaves similarly to standard Python and NumPy slicing.  
You can use row and column indices to extract specific parts of the DataFrame.

````python
# Example: Select Row 2, Columns 0 and 1
df.iloc[2, 0:2]
````

This selects the element(s) at row index `2` and column indices `0` and `1` (column `2` is excluded).

# Selecting specific elements with lists of positions

You can also pass **lists of integer positions** to `.iloc[]` to select **multiple rows and columns**.

````python
# Example: Select Rows 0 and 1, Columns 0 and 2
df.iloc[[0,1], [0,2]]
````

This selects a subset of the DataFrame:  
- Rows at positions 0 and 1  
- Columns at positions 0 and 2

For slicing rows explicitly:

````python
df.iloc[0:2,:]
````

# Slicing rows explicitly

You can use `:` to slice rows explicitly and select **all columns** by including `:` in the column position.

````python
df.iloc[:, 0:2]
````

This selects **all rows** and **columns 0 and 1**.

### Accessing a specific value

To access a **single value** in the DataFrame by its **row and column position**, use:

````python
df.iloc[0,0]
````

## Missing Data

In pandas, **missing data** is typically represented using `np.nan` (Not a Number),  
which comes from NumPy.

Missing values are **automatically excluded** from most computations (like `mean`, `sum`, etc.).

For more details, see the [Missing Data section](https://pandas.pydata.org/pandas-docs/stable/user_guide/missing_data.html#missing-data).

# Dropping Missing Data

The method `DataFrame.dropna()` removes any **rows** that contain at least one missing value (`NaN`).

This is useful when you want to work only with **complete cases** (no missing data).

````python
import numpy as np
import pandas as pd

data1 = [{'ID': '001', 'Age': np.nan, 'Sex':'M', 'CT':2.543}, {'Age': 56, 'ID': '002', 'Sex': 'F', 'CT':2.34}, {'Age': 56, 'ID': '003', 'Sex': 'F', 'CT':2.41}]
print (type(data1)) # List of dictionaries
print (type(data1[0])) # Each element is a dictionary

df1 = pd.DataFrame(data1)
print (df1)
````

This creates a DataFrame with one missing value (`NaN`) in the `'Age'` column of the first row.

````python
# Drop Rows with Any Missing Values (creates a new DataFrame)
df2 = df1.dropna(how="any")
print (df2) # Only rows without NaN are kept
print (df1) # The original DataFrame remains unchanged
````

````python
# Drop Rows with Any Missing Values (modifies the original DataFrame)
df1.dropna(how="any", inplace=True)
print(df1) # Now df1 has been updated: rows with NaN are removed
````

- `how="any"`: drop rows if **any column** has a missing value  
- `inplace=True`: perform the operation **in place** without returning a new DataFrame

## Stats

By default, pandas **excludes missing data** (`NaN`) from statistical operations.

You can compute common statistics such as **mean**, **sum**, **min**, **max**, etc.

````python
# Example: Mean of selected columns
df[['Age', 'CT']].mean()
````

This computes the mean only for the `'Age'` and `'CT'` columns.
