# Tuple

A **tuple** is a container that can store multiple values in a specific order.  
Unlike lists, tuples are **immutable**, which means their content cannot be changed after creation.

Tuples can contain elements of **different types** (e.g., integers, strings, etc.).

### Tuple with two integers

```python
tpl = (1, 2)
print(tpl)
print(type(tpl))
```

### Tuple are immutable: what it really means

Tuples are **immutable** — this means you **cannot change their individual elements** once the tuple is created.

---

### ❌ Not allowed: changing an element inside a tuple

```python
tpl = (1, 2)
tpl[0] = 5          # This will raise a TypeError
```

---

### ✅ Allowed: assigning a new tuple to the same variable (e.g., with mixed types)

```python
tpl = (1, 'Hello')
print(tpl)
print(type(tpl))
```

Here, you're **not modifying** the original tuple.  
You're just making the variable `tpl` point to a **new tuple** — this is a normal variable reassignment.

---

🧠 The tuple itself can't be changed, but the variable name can be reused for a different tuple.

### Tuple methods

The `tuple` class provides a few built-in methods, mostly for inspection.  
Since tuples are **immutable**, you cannot modify them, but you can still look inside them.

Here are two commonly used methods:
- `.count(value)` → returns how many times a value appears in the tuple  
- `.index(value)` → returns the index of the **first occurrence** of the value

---



```python
tpl = (1, 2, 3, 4, 6, 2, 3, 1, 4, 5)
print(tpl)

print(tpl.count(3))             # Return how many times 3 appears

print(tpl.index(4))             # Return the index of the first 4
```

### Tuples as function parameters

Tuples are often used as **function parameters** when you want to ensure that the values **cannot be modified** inside the function.

Since tuples are **immutable**, they protect the data from accidental changes during the function call.

This is useful when you want to **pass multiple values** together, but keep them safe from modification.

# Dictionary

A **dictionary** in Python is a collection of **key–value pairs**.  
Each key is unique and is used to access its corresponding value.

The syntax is: { key1: value1, key2: value2, … }

```python
mydict = {'CT': 'not performed', 'MRI': 3, 'ECG': 20}
print(mydict)
print(type(mydict))
```

In this example:
- `'CT'`, `'MRI'`, and `'ECG'` are the **keys**
- `'not performed'`, `3`, and `20` are the **values**

### Keys must be unique

In a dictionary, the **keys must be unique**.  
If you try to define the same key more than once, only the **last assigned value** will be kept.

```python
mydict = {'CT': 'not performed', 'CT': 3, 'ECG': 20}
print(mydict)
print(type(mydict))
```

In this example, the key `'CT'` was defined twice.  
Python keeps only the **last occurrence**.

### Accessing a dictionary

We can access the values in a dictionary by using their **unique keys**.

Dictionaries also provide helpful methods to explore their contents.

```python
print(1 in mydict)          # Checks if the key 1 exists in the dictionary
print(mydict.keys())        # Returns a list-like view of all keys

print(mydict['CT'])         # Accesses the value associated with the key 'CT'
```

- `in` checks for the existence of a key  
- `.keys()` returns all keys in the dictionary  
- Using square brackets `[]` with a key lets you retrieve its associated value

### Dictionaries are mutable

Dictionaries in Python are **mutable**, which means you can **add**, **modify**, or **remove** elements after the dictionary has been created.

```python
mydict['EEG']= [1, 2, 3]     # Adding a new (key, value) pair
print(mydict)

mydict['EEG']= 3             # Modifying the value associated with an existing key
print(mydict)
```

# Set

A **set** in Python is a collection of **unordered, unique** objects.  
If you try to add duplicate items, Python will automatically ignore them.

```python
st = {'CT', 'MRI'}
print(st)
print(type(st))
```

This creates a set containing two unique elements.  
Sets are defined using curly braces `{}`, just like dictionaries, but **without key–value pairs**.

### Set properties

Sets are **mutable**, so you can add or remove elements after creation.  
However, they are **unordered**, meaning the elements do not maintain any specific sequence.

```python
# Adding an element to a set
st.add('ECG')
print(st)
```

The `.add()` method adds a new element to the set.  
If the element is already present, nothing changes (no duplicates allowed).

```python
# Converting a list to a set
lst = ['CT', 'CT', 'MRI', 'ECG']
st = set(lst)
print(st) # Casting list to set removes duplicate items
```

This is a common way to remove duplicates from a list — by converting it into a set.
