# List

A **list** is a container that stores multiple elements in a specific order.  
Lists can hold objects of **different types** (e.g., numbers, strings, even other lists).

```python
lst = [0, 1, 2, "hi"]
print(lst)
print(type(lst))
```

# List methods

The `list` class provides many useful **methods** for working with ordered collections.  
You can add, remove, and sort elements.  

For a full list of list methods, see the [official Python documentation](https://docs.python.org/3/tutorial/datastructures.html).

The list class provides numerous methods (for a comprehensive list go to the [offical page](https://docs.python.org/3/tutorial/datastructures.html)) including:

```python
lst = ["Welcome", 2023]
print(lst)

lst.append('!')             # Add an item to the end of the list.
print(lst)

lst.insert(1, ',')          # Insert an item at index 1
print(lst)

lst.remove("Welcome")       # Remove the first occurrence of “Welcome”
print(lst)

lst = ['c', 'd', 'a']
print(lst)
lst.sort()                  # Sort the list in place (alphabetically)
print(lst)
```

Note: `.sort()` works only when all elements are of the **same, comparable type**.

### Lists are mutable

Lists are **mutable**, meaning their content can be **modified after creation**.  
You can access elements by their **index** and change them directly.

```python
lst = ['c', 'd', 'a']
print(lst)
lst[2] = 'e'
print(lst)
```

This replaces `'a'` with `'e'` at position 2 (remember: indexing starts from 0).
